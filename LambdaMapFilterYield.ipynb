```python
# 1. 매개변수 없음
hello = lambda: "안녕"

print(hello())  # 안녕
```

```python
# 2. 매개변수 1개 (일반 함수와 비교)
def add_one(x):
    return x + 1

add_one_lambda = lambda x: x + 1

print(add_one(3))         # 4
print(add_one_lambda(3))  # 4
```

```python
# 3. 매개변수 2개
add = lambda a, b: a + b

print(add(2, 3))  # 5
```

```python
# 4. 기본값 매개변수
greet = lambda name="손님": f"{name}님 안녕하세요"

print(greet())        # 손님님 안녕하세요
print(greet("철수"))  # 철수님 안녕하세요
```

```python
# 5. 조건 표현식
even_odd = lambda n: "짝수" if n % 2 == 0 else "홀수"

print(even_odd(4))  # 짝수
print(even_odd(7))  # 홀수
```

```python
# 6. 이름 없이 바로 호출
print((lambda x: x * 2)(5))  # 10
```

```python
# 7. 딕셔너리에 lambda 담기 (계산기)
calc = {
    "+": lambda a, b: a + b,
    "-": lambda a, b: a - b,
    "*": lambda a, b: a * b,
    "/": lambda a, b: a / b,
}

print(calc["+"](10, 2))  # 12
print(calc["-"](10, 2))  # 8
print(calc["*"](10, 2))  # 20
print(calc["/"](10, 2))  # 5.0
```

```python
# map
```

```python
# 1. 기본 형태
nums = [1, 2, 3, 4]
result = map(lambda x: x * 2, nums)

print(result)        # <map object at ...>
print(list(result))  # [2, 4, 6, 8]
```

```python
# 2. for 문과 비교
nums = [1, 2, 3, 4]

# for 문
result1 = []
for x in nums:
    result1.append(x * 2)

# map + lambda
result2 = list(map(lambda x: x * 2, nums))

print(result1)  # [2, 4, 6, 8]
print(result2)  # [2, 4, 6, 8]
```

```python
# 3. 제곱 구하기
nums = [1, 2, 3, 4, 5]

print(list(map(lambda x: x ** 2, nums)))  # [1, 4, 9, 16, 25]
```

```python
# 4. 문자열 처리
words = ["apple", "banana", "kiwi"]

print(list(map(lambda s: s.upper(), words)))  # ['APPLE', 'BANANA', 'KIWI']
```

```python
# 5. 조건 표현식
nums = [1, 2, 3, 4]

print(list(map(lambda n: "짝수" if n % 2 == 0 else "홀수", nums)))
# ['홀수', '짝수', '홀수', '짝수']
```

```python
# 6. 리스트 2개 동시 처리
a = [1, 2, 3]
b = [10, 20, 30]

print(list(map(lambda x, y: x + y, a, b)))  # [11, 22, 33]
```

```python
# 7. 튜플 리스트에서 값만 뽑기
scores = [("철수", 85), ("영희", 92), ("민수", 78)]

print(list(map(lambda s: s[0], scores)))  # ['철수', '영희', '민수']
print(list(map(lambda s: s[1], scores)))  # [85, 92, 78]
```

```python
# 8. map → 리스트
result = map(lambda x: x * 2, [1, 2, 3])

print(list(result))  # [2, 4, 6]
print(list(result))  # []  ← 비어 있음
```

```python
# 9. 내장 함수
text = "1 2 3"

print(text.split())                  # ['1', '2', '3']  (문자열)
print(list(map(int, text.split())))  # [1, 2, 3]        (정수)
```

```python
# filter
```

```python
# 1. 기본 형태
nums = [1, 2, 3, 4, 5]
result = filter(lambda x: x > 3, nums)

print(result)        # <filter object at ...>
print(list(result))  # [4, 5]
```

```python
# 2. for 문과 비교
nums = [1, 2, 3, 4, 5]

# for + if 문
result1 = []
for x in nums:
    if x > 3:
        result1.append(x)

# filter + lambda
result2 = list(filter(lambda x: x > 3, nums))

print(result1)  # [4, 5]
print(result2)  # [4, 5]
```

```python
# 3. 짝수만 고르기
nums = [1, 2, 3, 4, 5, 6]

print(list(filter(lambda x: x % 2 == 0, nums)))  # [2, 4, 6]
```

```python
# 4. 문자열 거르기
words = ["apple", "kiwi", "banana", "fig"]

print(list(filter(lambda w: len(w) >= 5, words)))  # ['apple', 'banana']
```

```python
# 5. 튜플 리스트 거르기
scores = [("철수", 85), ("영희", 92), ("민수", 78)]

print(list(filter(lambda s: s[1] >= 80, scores)))
# [('철수', 85), ('영희', 92)]
```

```python
# 6. 빈 값 제거
items = ["a", "", "b", ""]

print(list(filter(lambda s: s != "", items)))  # ['a', 'b']
```

```python
# 7. map과 비교
nums = [1, 2, 3, 4, 5]

print(list(map(lambda x: x > 3, nums)))     # [False, False, False, True, True]
print(list(filter(lambda x: x > 3, nums)))  # [4, 5]
```

```python
# 8. filter + map 함께
nums = [1, 2, 3, 4, 5, 6]

evens = filter(lambda x: x % 2 == 0, nums)  # 짝수만 고르고
squares = map(lambda x: x ** 2, evens)      # 제곱

print(list(squares))  # [4, 16, 36]
```

```python
# yield
```

```python
# 1. yield 하나
def say_hello():
    yield "안녕"

gen = say_hello()

print(gen)        # <generator object say_hello at ...>
print(next(gen))  # 안녕
```

```python
# 2. yield 여러 개
def count_three():
    yield 1
    yield 2
    yield 3

gen = count_three()

print(next(gen))  # 1
print(next(gen))  # 2
print(next(gen))  # 3
# print(next(gen))  # 더 꺼낼 값이 없으면 StopIteration 예외 발생
```

```python
# 3. 멈췄다가 이어서 실행
def steps():
    print("1단계 실행")
    yield "A"
    print("2단계 실행")
    yield "B"
    print("3단계 실행")
    yield "C"

gen = steps()  # 실행 전

print(next(gen))
# 1단계 실행
# A

print(next(gen))
# 2단계 실행
# B

print(next(gen))
# 3단계 실행
# C
```

```python
# 4. for 문으로 꺼내기
# for 문이 next()를 대신 불러 주고, 값이 다 떨어지면 에러 없이 알아서 멈춤
def count_three():
    yield 1
    yield 2
    yield 3

for value in count_three():
    print(value)
# 1
# 2
# 3
```

```python
# 5. list()로 한 번에 꺼내기
def count_three():
    yield 1
    yield 2
    yield 3

gen = count_three()

print(list(gen))  # [1, 2, 3]
print(list(gen))  # []  ← 이미 다 꺼내서 비어 있음
```

```python
# 6. 조건과 함께
def even_only(nums):
    for x in nums:
        if x % 2 == 0:
            yield x  # 짝수일 때만

print(list(even_only([1, 2, 3, 4, 5, 6])))  # [2, 4, 6]
```

```python
# 7. return과 비교

# return
def even_return(nums):
    result = []
    for x in nums:
        if x % 2 == 0:
            result.append(x)
    return result

# yield
def even_yield(nums):
    for x in nums:
        if x % 2 == 0:
            yield x

nums = [1, 2, 3, 4, 5, 6]

print(even_return(nums))        # [2, 4, 6]
print(even_yield(nums))         # <generator object even_yield at ...>
print(list(even_yield(nums)))   # [2, 4, 6]
```
