```python
# 가장 기본적인 클래스: 속성도 메서드도 없음

class Dog:
    pass

d1 = Dog()
d2 = Dog()   # 클래스로 객체(인스턴스)를 여러 번 만들 수 있음

print(d1)
print(d2)
```

```python
# 같은 클래스로 만들었어도 각 객체는 서로 다른 별개의 객체

print(d1 is d2)
print(type(d1))
```

```python
# 괄호 없이 클래스 이름만 쓰면 객체 생성이 아니라 클래스 자체를 가리킴

print(Dog)
print(type(Dog))
```

```python
# __init__: 객체가 만들어질 때 자동으로 실행되는 초기화 메서드

class Person:
    def __init__(self, name, age):
        self.name = name   # 인스턴스 속성
        self.age = age

p = Person("철수", 20)
print(p.name)
print(p.age)
```

```python
# 객체마다 자신만의 속성 값을 가짐

p1 = Person("영희", 22)
p2 = Person("민수", 25)

print(p1.name, p1.age)
print(p2.name, p2.age)
```

```python
# 객체를 만들 때도 위치 인자와 키워드 인자를 쓸 수 있음

p3 = Person("지영", 30)
p4 = Person(name="동현", age=28)
p5 = Person(age=27, name="수진")   # 키워드 인자는 순서를 바꿔도 됨

print(p3.name, p4.name, p5.name)
```

```python
# 속성은 객체를 만든 뒤에도 바꿀 수 있음

p1.age = 23
print(p1.age)
print(p2.age)   # 다른 객체는 영향을 받지 않음
```

```python
# 메서드: 클래스 안에 정의한 함수, 첫 번째 매개변수는 self

class Counter:
    def __init__(self):
        self.count = 0

    def increase(self):
        self.count += 1

    def get(self):
        return self.count

c = Counter()
c.increase()
c.increase()
print(c.get())
```

```python
# self는 메서드를 호출한 객체 자신을 가리킴
# c.increase()는 Counter.increase(c)와 같음

Counter.increase(c)
print(c.get())
```

```python
# 메서드도 매개변수와 반환값을 가질 수 있음

class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount > self.balance:
            return "잔액 부족"   # return을 만나면 메서드가 즉시 종료됨
        self.balance -= amount
        return self.balance

acc = Account("철수", 10000)
print(acc.deposit(5000))
print(acc.withdraw(3000))
print(acc.withdraw(50000))
```

```python
# 메서드 안에서 같은 객체의 다른 메서드를 호출할 때도 self를 사용

class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def describe(self):
        print(f"{self.width} x {self.height} 사각형, 넓이 {self.area()}")

r = Rectangle(3, 4)
r.describe()
```

```python
# 클래스 속성: 모든 객체가 공유 / 인스턴스 속성: 객체마다 따로

class Student:
    school = "한빛고등학교"   # 클래스 속성

    def __init__(self, name):
        self.name = name     # 인스턴스 속성

s1 = Student("지영")
s2 = Student("동현")

print(s1.name, s1.school)
print(s2.name, s2.school)
print(Student.school)   # 클래스 속성은 클래스 이름으로도 접근 가능
```

```python
# 클래스 속성을 바꾸면 모든 객체에 반영됨

Student.school = "새솔고등학교"
print(s1.school)
print(s2.school)
```

```python
# 객체를 통해 대입하면 클래스 속성이 바뀌는 게 아니라
# 그 객체에만 같은 이름의 인스턴스 속성이 새로 생김

s1.school = "다른고등학교"
print(s1.school)
print(s2.school)
print(Student.school)
```

```python
# 클래스 함수(@classmethod): 첫 번째 매개변수 cls는 클래스 자신을 가리킴

class Member:
    count = 0   # 만들어진 회원 수 (클래스 속성)

    def __init__(self, name):
        self.name = name
        Member.count += 1

    @classmethod
    def get_count(cls):
        return cls.count

m1 = Member("철수")
m2 = Member("영희")

print(Member.get_count())   # 클래스 이름으로 호출
print(m1.get_count())       # 객체로 호출해도 cls에는 클래스가 들어감
```

```python
# 클래스 함수로 클래스 속성 바꾸기

class Shop:
    discount = 10   # 할인율(%)

    def __init__(self, item, price):
        self.item = item
        self.price = price

    def final_price(self):
        return self.price * (100 - Shop.discount) / 100

    @classmethod
    def set_discount(cls, rate):
        cls.discount = rate

a = Shop("연필", 1000)
b = Shop("공책", 3000)
print(a.final_price(), b.final_price())

Shop.set_discount(50)   # 모든 객체에 반영됨
print(a.final_price(), b.final_price())
```

```python
# 클래스 함수로 객체를 만드는 다른 방법 제공하기

class Date:
    def __init__(self, year, month, day):
        self.year = year
        self.month = month
        self.day = day

    @classmethod
    def from_string(cls, text):
        # "2026-10-06" 형태의 문자열을 나눠서 객체를 만들어 반환
        year, month, day = text.split("-")
        return cls(int(year), int(month), int(day))

d = Date.from_string("2026-10-06")
print(d.year, d.month, d.day)
```

```python
# __str__이 없으면 print로 객체를 출력했을 때 알아보기 어려움

class Book:
    def __init__(self, title, price):
        self.title = title
        self.price = price

b = Book("파이썬 입문", 25000)
print(b)
```

```python
# __str__: print로 객체를 출력할 때 보여줄 문자열을 return으로 돌려줌

class Book:
    def __init__(self, title, price):
        self.title = title
        self.price = price

    def __str__(self):
        return f"{self.title} ({self.price}원)"

b = Book("파이썬 입문", 25000)
print(b)
print(str(b))   # str()로 바꿀 때도 __str__이 사용됨
```

```python
# 상속: 기존 클래스(부모)의 속성과 메서드를 물려받아 새 클래스(자식)를 만듦

class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name}이(가) 먹는다")

class Cat(Animal):   # 괄호 안에 부모 클래스를 적음
    pass

cat = Cat("나비")
print(cat.name)   # 부모의 __init__이 그대로 사용됨
cat.eat()         # 부모의 메서드를 그대로 사용
```

```python
# 자식 클래스에 새로운 메서드를 추가할 수 있음

class Bird(Animal):
    def fly(self):
        print(f"{self.name}이(가) 난다")

bird = Bird("참새")
bird.eat()   # 부모에게 물려받은 메서드
bird.fly()   # 자식에서 추가한 메서드

# 부모 클래스의 객체는 자식이 추가한 메서드를 쓸 수 없음
animal = Animal("동물")
animal.eat()
# animal.fly()   # AttributeError: 'Animal' object has no attribute 'fly'
```

```python
# 자식 클래스의 객체는 부모 클래스의 객체로도 인정됨

print(isinstance(bird, Bird))
print(isinstance(bird, Animal))
print(isinstance(animal, Bird))
```
