```python
# 딕셔너리 생성
student = {"name": "철수", "age": 20, "major": "컴퓨터공학"}
print(student)

# 값 접근
print(student["name"])
print(student.get("age"))
```

```python
scores = {"math": 90, "english": 85, "science": 78}

for key in scores:
    print(key, scores[key])

for key, value in scores.items():
    print(f"{key}: {value}점")
```

```python
inventory = {"apple": 10, "banana": 0, "orange": 5}

for item, count in inventory.items():
    if count == 0:
        print(f"{item}은(는) 품절입니다.")
    else:
        print(f"{item}: {count}개 남음")

print("apple" in inventory)
print("grape" in inventory)
```

```python
numbers = [1, 2, 3, 4, 5]
squared = {n: n**2 for n in numbers}
print(squared)

# 조건 포함
even_squared = {n: n**2 for n in numbers if n % 2 == 0}
print(even_squared)
```

```python
# 튜플

# 괄호 O, 콤마 O, 요소 여러 개 → 튜플
a = (1, 2, 3)
print(a, type(a))

# 괄호 X, 콤마 O, 요소 여러 개 → 튜플
b = 1, 2, 3
print(b, type(b))

# 괄호 O, 콤마 O, 요소 1개 → 튜플
c = (1,)
print(c, type(c))

# 괄호 X, 콤마 O, 요소 1개 → 튜플
d = 1,
print(d, type(d))

# 괄호 O, 콤마 X → 튜플 아님 (그냥 값)
e = (1)
print(e, type(e))

# 괄호 X, 콤마 X → 튜플 아님 (그냥 값)
f = 1
print(f, type(f))

# 빈 괄호, 콤마 없음 → 튜플 (유일하게 콤마 없이도 튜플이 되는 경우)
g = ()
print(g, type(g))

# 콤마만 있고 괄호 없이 "빈" 튜플은 만들 수 없음 -> 문법 자체가 성립 안 함
# h = ,  # SyntaxError
```

```python
# 기본 언패킹
point = (3, 5)
x, y = point
print(x, y)

# 여러 변수 동시 할당도 사실 튜플 언패킹
a, b, c = 1, 2, 3
print(a, b, c)

# 변수 교환 (swap)
a, b = b, a
print(a, b)

# 함수의 여러 반환값도 언패킹으로 받기
def get_coordinates():
    return 10, 20

x, y = get_coordinates()
print(x, y)

# 개수가 안 맞으면 에러
try:
    x, y = (1, 2, 3)
except ValueError as e:
    print("에러 발생:", e)
```
