```python
# 가장 기본적인 함수: 매개변수도 반환값도 없음

def say_hello():
    print("안녕하세요!")

say_hello()
say_hello()   # 함수는 여러 번 호출할 수 있음
```

```python
# 함수를 정의만 하고 호출하지 않으면 아무 일도 일어나지 않음

def say_bye():
    print("안녕히 가세요!")

# say_bye()  <- 호출하지 않았으므로 출력 없음
```

```python
# 괄호 없이 함수 이름만 쓰면 호출이 아니라 함수 객체 자체를 가리킴

print(say_hello)
print(type(say_hello))
```

```python
# 호출은 반드시 괄호가 필요

say_bye()
```

```python
# 매개변수가 있고 return으로 값을 돌려주는 함수

def add(a, b):
    return a + b

result = add(3, 5)
print(result)
```

```python
# 반환값을 변수에 저장하지 않고 바로 사용

print(add(10, 20))
print(add(1, 2) + add(3, 4))
```

```python
# 함수 안에서 다른 함수를 호출

def square(n):
    return n * n

def sum_of_squares(a, b):
    return square(a) + square(b)

print(sum_of_squares(3, 4))
```

```python
# 같은 함수를 다양한 타입에 사용 (파이썬은 타입을 강제하지 않음)

print(add(1.5, 2.5))
print(add("안녕", "하세요"))
print(add([1, 2], [3, 4]))
```

```python
# return을 만나면 함수가 즉시 종료됨

def check(n):
    if n < 0:
        return "음수"
    print("음수 검사 통과")
    return "0 이상"

print(check(-1))
print(check(5))
```

```python
# 용어 정리: 매개변수(parameter)는 함수 정의 쪽, 인자(argument)는 호출 쪽

# name, age, city: 매개변수
def introduce(name, age, city):
    print(f"{name}, {age}세, {city} 거주")
```

```python
# 1) 위치 인자: 순서대로 매개변수에 대응
introduce("철수", 20, "서울")

# 위치 인자는 순서가 바뀌면 의미도 바뀜
introduce("서울", 20, "철수")

# 2) 키워드 인자: 매개변수 이름을 지정해서 전달
introduce(name="영희", age=22, city="부산")

# 키워드 인자는 순서를 바꿔도 됨
introduce(city="대전", name="민수", age=25)

# 3) 위치 인자와 키워드 인자 섞기: 위치 인자가 반드시 먼저
introduce("지영", 30, city="인천")
introduce("동현", age=28, city="광주")

# 위치 인자를 키워드 인자 뒤에 쓰면 문법 에러 (아래 줄은 에러 발생)
# introduce(name="철수", 20, "서울")   # SyntaxError: positional argument follows keyword argument
```

```python
def calc_price(price, quantity, discount):
    # 가격 * 수량에 할인율(%)을 적용한 최종 금액을 반환
    total = price * quantity
    return total * (100 - discount) / 100

# 위치 인자만: 무슨 값인지 알아보기 어려움
print(calc_price(5000, 3, 10))

# 키워드 인자만: 의미가 분명해서 읽기 쉬움
print(calc_price(price=5000, quantity=3, discount=10))

# 키워드 인자는 순서를 바꿔도 결과가 같음
print(calc_price(discount=10, price=5000, quantity=3))

# 섞어 쓰기: 앞쪽에 위치, 뒤쪽은 키워드
print(calc_price(5000, 3, discount=10))
print(calc_price(5000, quantity=3, discount=10))
```

```python
# return이 없는 함수는 None을 반환
def greet(name):
    print(f"안녕하세요, {name}님!")

result = greet("철수")
print(result)
print(type(result))
```

```python
# return만 쓰고 값을 주지 않아도 None
def do_nothing():
    return

print(do_nothing())
```

```python

# 조건에 따라 return이 실행되지 않으면 None
def get_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    # 80점 미만이면 return을 만나지 않고 함수가 끝남

print(get_grade(95))
print(get_grade(85))
print(get_grade(50))   # None
```

```python
# None인지 확인할 때는 is None을 사용
grade = get_grade(50)
if grade is None:
    print("등급 없음")
```

```python
# print와 return의 차이
def add_print(a, b):
    print(a + b)       # 화면에 출력만 함

def add_return(a, b):
    return a + b       # 값을 돌려줌

x = add_print(1, 2)    # 3이 출력되고
print(x)               # x는 None

y = add_return(1, 2)   # 아무것도 출력되지 않고
print(y)               # y는 3
```

```python
print(add_return(1, 2) * 10)    # 반환값은 계산에 사용 가능
```
