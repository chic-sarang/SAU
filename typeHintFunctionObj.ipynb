```python
# 6. 타입 힌트는 강제되지 않음
def add(a: int, b: int) -> int:
    return a + b

print(add(2, 3))             # 5
print(add("안녕", "하세요"))  # 안녕하세요  ← int가 아닌데도 에러 없이 실행
print(add(1.5, 2.5))         # 4.0
```

```python
# 8. 튜플
student: tuple[str, int] = ("철수", 85)   # 개수 고정: 첫 번째는 str, 두 번째는 int
nums: tuple[int, ...] = (1, 2, 3, 4)      # 개수 자유: 모두 int

print(student)  # ('철수', 85)
print(nums)     # (1, 2, 3, 4)
```

```python
# 9. 딕셔너리
scores: dict[str, int] = {"철수": 85, "영희": 92}  # 키는 str, 값은 int

def get_score(data: dict[str, int], name: str) -> int:
    return data[name]

print(get_score(scores, "영희"))
```

```python
# 11. 둘 중 하나
def show_id(user_id: int | str) -> str:
    return f"ID: {user_id}"

print(show_id(1001))     # ID: 1001
print(show_id("A-01"))   # ID: A-01
```

```python
# 12. None 허용
def find_score(data: dict[str, int], name: str) -> int | None:
    if name in data:
        return data[name]
    return None  # 없으면 None

scores = {"철수": 85, "영희": 92}

print(find_score(scores, "철수"))  # 85
print(find_score(scores, "민수"))  # None
```

```python
# 13. 아무 타입이나
from typing import Any

def show(value: Any) -> None:
    print(value, type(value))

show(10)        # 10 <class 'int'>
show("안녕")    # 안녕 <class 'str'>
show([1, 2])    # [1, 2] <class 'list'>
```

```python
# 2. None의 타입
print(type(None))  # <class 'NoneType'>
print(type(10))    # <class 'int'>
print(type("안녕")) # <class 'str'>
```

```python
# 3. None 비교는 is로
x = None

print(x is None)      # True
print(x is not None)  # False
```

```python
# 4. 반환값이 없는 함수는 None을 리턴
def say_hello():
    print("안녕")

result = say_hello()  # 안녕

print(result)  # None
```

```python
# 6. None은 False처럼 동작
print(bool(None))  # False

x = None
if not x:
    print("값이 없어요")

# 주의: 0과 ""도 False처럼 동작
# None인지 확인할 때는 is None을 사용
a = 0
b = ""

print(bool(a), a is None)  # False False
print(bool(b), b is None)  # False False
```

```python
# 7. 기본값으로 None 쓰기
# 타입 힌트에서는 NoneType 대신 None이라고 씀

def greet(name: str | None = None) -> str:
    if name is None:
        return "손님님 안녕하세요"
    return f"{name}님 안녕하세요"

print(greet())        # 손님님 안녕하세요
print(greet("철수"))  # 철수님 안녕하세요
```

```python
# 1. 함수를 변수에 담기
def hello():          # 이 줄이 실행될 때 function 클래스의 객체가 만들어지고,
    print("안녕")      # hello라는 이름이 붙여짐 (아직 "안녕"은 출력되지 않음)

say = hello  # 괄호 없이 쓰면 위에서 만들어진 함수 객체를 말함

say()    # 안녕  ← ()를 붙이면 실행
hello()  # 안녕

print(type(hello))    # <class 'function'> ← 함수는 function 클래스의 객체예요
print(say.__name__)   # hello              ← 원래 이름도 그대로
print(say is hello)   # True               ← 둘은 같은 객체
```

```python
# 2. 함수를 다른 함수에 넘기기
def double(x):
    return x * 2

def square(x):
    return x ** 2

def apply(func, value):
    return func(value)  # 넘겨받은 함수를 실행

print(apply(double, 5))  # 10
print(apply(square, 5))  # 25
```

```python
# 3. 함수를 돌려주기
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

def choose(op):
    if op == "+":
        return add  # 함수 자체를 돌려줌
    return sub

f = choose("+")
print(f(10, 3))  # 13

g = choose("-")
print(g(10, 3))  # 7
```
