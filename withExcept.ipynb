```python
class MyContext:
    def __enter__(self):
        print("with 블록 진입")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("with 블록 종료")


with MyContext() as obj:
    print("with 블록 내부 실행")
```

```python
# 예외가 발생해도 exit() 실행

with MyContext() as obj:
    print("작업 시작")
    result = 10 / 0
    print("작업 종료")
```

```python
class Study:
    def __enter__(self):
        print("공부를 시작합니다.")
        return self

    def __exit__(self, exc_type, exc_value, traceback):
        print("공부를 마칩니다.")


with Study():
    print("파이썬 문법 공부 중...")
    print("예제 코드 작성 중...")

print("다음 수업으로 이동합니다.")
```

```python
# 예외가 발생했는데 이를 처리하지 않으면 프로그램의 정상적인 실행이 중단됨
# 예외를 처리한다는 말은 "프로그램 실행 중 문제가 발생했을 때, 프로그램이 멈추지 않도록 적절한 조치를 취하는 것"
```

```python
try:
    pass
    # 예외가 발생할 가능성이 있는 코드

except:
    pass
    # 예외가 발생했을 때 실행할 코드
```

```python
try:
    result = 10 / 0         # 예외 발생
    print(result)           # 이 부분은 실행되지 않음

except:
    print("오류가 발생했습니다.")

print("프로그램 흐름은 계속됨")
```

```python
# 1. 0으로 나누기 - 예외 처리
try:
    result = 10 / 0
    print(result)
except:
    print("0으로 나눌 수 없습니다.")

print("프로그램 흐름은 계속됨")
```

```python
# 2. 정의되지 않은 변수 사용 - 예외 처리
try:
    print(user_name)
except:
    print("정의되지 않은 변수입니다.")

print("프로그램 흐름은 계속됨")
```

```python
# 3. 잘못된 자료형 간 연산 - 예외 처리
try:
    result = "100" + 50
    print(result)
except:
    print("서로 다른 자료형은 직접 더할 수 없습니다.")

print("프로그램 흐름은 계속됨")
```

```python
# 4. 존재하지 않는 리스트 인덱스 접근 - 예외 처리
try:
    numbers = [10, 20, 30]
    print(numbers[5])
except:
    print("존재하지 않는 인덱스입니다.")

print("프로그램 흐름은 계속됨")
```

```python
# 5. 존재하지 않는 파일 열기 - 예외 처리
try:
    f = open("없는파일.txt", "r")
except:
    print("파일을 찾을 수 없습니다.")

print("프로그램 흐름은 계속됨")
```

```python
# except 뒤에 예외 종류를 지정하면 해당 예외가 발생했을 때만 실행 - 1

try:
    result = 10 / 0

except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")

print("프로그램 종료")
```

```python
# except 뒤에 예외 종류를 지정하면 해당 예외가 발생했을 때만 실행 - 2

try:
    number = int(input("숫자를 입력하세요: "))
    result = 100 / number
    print("계산 결과:", result)

except ValueError:
    print("숫자만 입력해야 합니다.")

except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")

print("프로그램 종료")
```

```python
# 여러 예외를 하나의 except로 처리하기 - 3

try:
    number = int(input("숫자를 입력하세요: "))
    result = 100 / number
    print("계산 결과:", result)

except (ValueError, ZeroDivisionError):
    print("입력값을 확인하세요.")

print("프로그램 종료")
```

```python
# else는 예외가 발생하지 않았을 때 실행되는 블록

try:
    result = 10 / 2
    print("계산 결과:", result)

except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")

else:
    print("계산이 정상적으로 완료되었습니다.")

print("프로그램 종료")
```

```python
# finally는 예외 발생 여부와 관계없이 항상 실행되는 블록

try:
    result = 10 / 0
    print("계산 결과:", result)

except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")

finally:
    print("계산 작업을 마쳤습니다.")

print("프로그램 종료")
```

```python
# 예외 정보 확인하기 - 예외 메시지 확인하기

try:
    result = 10 / 0

except ZeroDivisionError as e:
    print("예외가 발생했습니다.")
    print("예외 내용:", e)

print("프로그램 종료")
```

```python
# 일부러 예외 발생시키기 - raise 예외종류("예외 메시지")

age = int(input("나이를 입력하세요: "))

if age < 0:
    raise ValueError("나이는 음수일 수 없습니다.")

print("입력한 나이:", age)
```

```python
# 일부러 예외 발생시키기 - 발생된 예외 처리하기

try:
    age = int(input("나이를 입력하세요: "))

    if age < 0:
        raise ValueError("나이는 음수일 수 없습니다.")

    print("입력한 나이:", age)

except ValueError as e:
    print("입력 오류:", e)

print("프로그램 종료")
```
